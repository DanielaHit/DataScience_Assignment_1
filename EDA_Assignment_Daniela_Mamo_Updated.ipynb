### File & Technical Overview

* **Collection Intent:** Research-oriented psychometric dataset designed to observe adult attachment dynamics, childhood emotional markers, and relational satisfaction outcomes.

### Structural & Column Breakdown
The dataset incorporates demographic factors, continuous psychological rating scales, and categorical behavior indicators:

* **Demographic Baseline:** `age` (Continuous numeric integer), `gender` (Categorical variable), `relationship_status` (Categorical variable), `parenting_style` (Categorical background variable).
* **Psychometric & Relational Scales:** `comfort_with_intimacy`, `fear_of_abandonment`, `trust_in_others`, `openness_in_relationships`, `social_support_perception`, `relationship_satisfaction` (Discrete 1–5 integer Likert scales).

### Potential Biases & Methodological Limitations
1. **Self-Report Bias:** All behavioral and relational scores rely entirely on participant self-evaluations. Respondents may overestimate positive metrics (such as trust or intimacy comfort) while underreporting anxiety or abandonment fears.
2. **Social Desirability Bias:** Sensitive self-assessments regarding relationship satisfaction and mental health status are susceptible to social desirability, where participants select responses aligned with perceived societal norms.
3. **Cross-Sectional Limitation:** The data provides a single temporal snapshot, limiting the ability to infer direct long-term longitudinal causality in how attachment styles evolve over time.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('attachment_survey_dummy_3000.csv')
print(f"Shape: {df.shape}")
print(df.info())
```

```python
# Missing values & duplicates
print("Missing values:\n", df.isnull().sum()[df.isnull().sum() > 0])
print("Duplicates:", df.duplicated().sum())

# Univariate stats & Outliers (3 methods for numerical columns)
num_cols = ['age', 'comfort_with_intimacy', 'fear_of_abandonment', 'trust_in_others', 'relationship_satisfaction']
for col in num_cols:
    s = df[col]
    q25, q75 = s.quantile(0.25), s.quantile(0.75)
    iqr = q75 - q25
    z = (s - s.mean()) / s.std()

    out_z = (abs(z) > 3).sum()
    out_iqr = ((s < q25 - 1.5 * iqr) | (s > q75 + 1.5 * iqr)).sum()
    out_p = ((s < s.quantile(0.01)) | (s > s.quantile(0.99))).sum()

    print(f"{col} | Mean: {s.mean():.2f}, Std: {s.std():.2f} | Outliers (Z/IQR/Pct): {out_z}/{out_iqr}/{out_p}")
```

#### Outliers Analysis Discussion
By comparing three different outlier detection methods (Z-score, IQR, and Percentiles), we observed distinct behaviors in our dataset. The IQR method was more sensitive to extreme psychological scores (especially in *fear_of_abandonment* and *relationship_satisfaction*), revealing that a subset of participants experience heightened emotional anxiety that deviates from the normal distribution.

```python
categorical_cols = ['gender', 'relationship_status', 'parenting_style', 'conflict_response', 'mental_health_status', 'attachment_style_label']

for col in categorical_cols:
    print(f"\n==========================================")
    print(f" FREQUENCY TABLE: '{col}'")
    print(f"==========================================")

    counts = df[col].value_counts(dropna=False)
    percentages = (df[col].value_counts(normalize=True, dropna=False) * 100).round(2)

    freq_table = pd.DataFrame({'Count': counts, 'Percentage (%)': percentages})
    print(freq_table)

    rare = percentages[percentages < 5.0]
    if not rare.empty:
        print(f"\n[Warning] Rare categories detected (<5%): {list(rare.index)}")
    else:
        print("\nNo rare categories detected (<5% threshold).")
```

### 5.2 Categorical Variables & Rare Category Handling

#### Frequency Distribution Overview
* **Value Counts & Proportions:** Absolute frequencies and percentage distributions were calculated across all categorical columns (`gender`, `parenting_style`, `conflict_response`, `mental_health_status`, and `attachment_style_label`).
* **Dominant Classes:** Primary categories retain balanced distributions, capturing core demographic and psychological typologies across the $N=3,000$ sample.

#### Strategy for Rare Categories
* **Threshold & Identification:** Categories accounting for less than 5% of the total dataset are flagged as rare categories.
* **Handling Method:**
  1. **Informational Retention (EDA Phase):** During exploratory analysis, rare responses are retained to observe unique qualitative signals without distorting base distributions.
  2. **Grouping / Binning (Pre-Modeling Phase):** If utilized for machine learning feature engineering, categories below the 5% threshold will be grouped into a single consolidated `'Other'` category. This prevents statistical noise, reduces dummy variable cardinality, and avoids overfitting.

```python
import scipy.stats as stats

print("=== 7. BONUS SECTION ANALYSIS ===")

# 1. Time Dependency Check
df['row_index'] = range(len(df))
time_corr, time_pvalue = stats.pearsonr(df['row_index'], df['relationship_satisfaction'])
print(f"1. Time Dependency Check: Correlation = {time_corr:.4f}, p-value = {time_pvalue:.4f}")

# 2. Engineering Feature
df['emotional_wellbeing_index'] = (
    df['comfort_with_intimacy'] +
    df['trust_in_others'] +
    (6 - df['fear_of_abandonment'])
) / 3
print(f"2. Engineering Feature Created: 'emotional_wellbeing_index' (Mean: {df['emotional_wellbeing_index'].mean():.2f})")

# 3. Hypothesis Testing
groups = [group['relationship_satisfaction'].values for name, group in df.groupby('attachment_style_label')]
anova_f, anova_p = stats.f_oneway(*groups)
print(f"3. Hypothesis Testing (ANOVA on Relationship Satisfaction by Attachment Style): F-stat = {anova_f:.4f}, p-value = {anova_p:.4f}")

# 4. Suggestions for Future Research
print("\n4. Suggestions for Future Research:")
print("- Expanding the sample size across diverse demographic groups and age ranges.")
print("- Longitudinal tracking of attachment styles over time during major life transitions.")
print("- Integrating qualitative interviews alongside quantitative metrics to capture deeper emotional nuances.")


# ==========================================
# 8. CONCLUSION & MEASUREMENT REFLECTIONS
# ==========================================
print("\n=== 8. CONCLUSION & MEASUREMENT REFLECTIONS ===")
print("Measurement Focus:")
print("- Data is not 'Truth' — it is the product of a measurement process.")
print("Key questions to consider:")
print("  * What was measured?")
print("  * What is missing?")
print("  * What is biased?")
print("  * What was artificially 'centered'?")
```
